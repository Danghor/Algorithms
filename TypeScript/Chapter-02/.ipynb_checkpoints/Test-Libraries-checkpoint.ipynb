# Test your libraries

Let us begin by checking each library, you will need for this course.

First we are testing if fraction.js was installed correctly by adding 1/2 and 1/3.
Expected output: `Fraction: 5/6`

```typescript
import Fraction from 'fraction.js';

const result = new Fraction(1, 2).add(new Fraction(1, 3));
console.log("Fraction:", result.toFraction());
```

Next we test whether `@hpcc-js/wasm` is installed correctly by loading the Graphviz engine and rendering a small graph to SVG.

```typescript
import * as tslab from "tslab";
import { Graphviz } from '@hpcc-js/wasm';

const gv  = await Graphviz.load();
const dot = 'digraph G { A -> B }';
const svg = gv.layout(dot, "svg", "dot");
tslab.display.svg(svg)
```

Next we  test if `mathjs` is installed correctly by creating and displaying a simple color gradient.

```typescript
import { complex, add } from 'mathjs';
const a = complex(2,  3);          // 2 + 3i
const b = complex(4, -1);          // 4 - 1i
console.log(add(a, b).toString()); // Outputs: 6 + 2i
```

Next we  test if `pngjs` is installed correctly by creating and displaying a simple color gradient.

```typescript
import { PNG } from 'pngjs';
import * as tslab from 'tslab';

// 1. Create a blank 256x256 canvas in memory
const width  = 256;
const height = 256;
const png    = new PNG({ width, height });

// 2. Mathematically generate a gradient
for (let y = 0; y < png.height; y++) {
    for (let x = 0; x < png.width; x++) {
        let idx = (png.width * y + x) << 2; // Calculate buffer index
        // Fill in the RGBA values
        png.data[idx] = x;           // Red increases from left to right
        png.data[idx+1] = y;         // Green increases from top to bottom
        png.data[idx+2] = 255 - x;   // Blue decreases from left to right
        png.data[idx+3] = 255;       // Alpha (255 = fully opaque)
    }
}

// 3. Convert to a base64 string
const outputBuffer = PNG.sync.write(png);
const base64Image  = outputBuffer.toString('base64');

// 4. Render it in Jupyter!
const html = `<img src="data:image/png;base64,${base64Image}" />`;
tslab.display.html(html);
```

Next we test whether heap-js is installed correctly by creating a min-heap and finding the smallest number.
Expected output: `1,5,100,10`

```typescript
import { Heap } from 'heap-js';

const minHeap = new Heap<number>();
minHeap.push(10, 5, 100, 1);
minHeap.toString()
```

Next we test whether `logic-solver` is installed correctly by solving a simple SAT problem (A or B, but not A).
Expected output: `Logic-Solver (B is true?): true`

```typescript
import Logic from 'logic-solver';

const logicSolver = new Logic.Solver();
logicSolver.require(Logic.or("A", "B"));
logicSolver.require(Logic.not("A"));

const solution = logicSolver.solve();

if (solution !== null) {
    console.log("Logic-Solver (B is true?):", solution.evaluate("B"));
} else {
    console.log("No solution found");
}
```

Finally, we test whether `z3-solver` is installed correctly by finding an integer `x` such that $x^2 + 3 \cdot x - 4 = 0$ and $x < 0$.

```typescript
import { init } from 'z3-solver';

const { Context } = await init();
const { Solver, Int } = Context("main");
const z3 = new Solver();
const x  = Int.const('x');

z3.add(x.mul(x).add(x.mul(3)).sub(4).eq(0), x.lt(0));
await z3.check()
```

```typescript
z3.model().eval(x).toString()
```
