# Memoizing the Fibonacci Sequence in TypeScript


We use a higher-order function to wrap the recursive calls.

```typescript
function memoizeFunction<T extends (...args: any[]) => any>(fn: T): T {
    const cache = new Map<string, any>();
    
    return function(...args: any[]) {
        // Apply the same replacer here
        const key = JSON.stringify(args, (_, value) => 
            typeof value === 'bigint' ? value.toString() : value
        );       
        if (cache.has(key)) {
            return cache.get(key);
        }
        const result = fn(...args);
        cache.set(key, result);
        return result;
    } as T;
}
```

We must assign the memoized version to the variable `fibonacciFunctional` 
so the recursive calls within it also utilize the cache.

```typescript
const fibonacci = memoizeFunction((n: bigint): bigint => {
    if (n <= 1n) {
        return n;
    }
    return fibonacci(n - 1n) + fibonacci(n - 2n);
});
```

```typescript
for (let i = 0n; i <= 100n; ++i) {
    console.log(`fibonacci(${i}) = ${fibonacci(i)}`);
} 
```

```typescript
console.time("Memoized Function");
console.log(`fibonacci(1000n) = ${fibonacci(1000n)}`);
console.timeEnd("Memoized Function");
```
